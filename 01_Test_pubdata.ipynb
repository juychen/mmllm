```python
dfh_cov = pd.read_csv('/data1st1/junyi/methdata/GSE244251/GSE244251_Methylation_5hmC_sites_cov9.bed.gz', sep='\t')
dfm_cov = pd.read_csv('/data1st1/junyi/methdata/GSE244251/GSE244251_Methylation_5mC_sites_cov9.bed.gz', sep='\t')
```

```python
output_dir = '/data1st1/junyi/methdata/GSE244251/processed_data'
import os
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
dfm_cov.head()
```

```python
# 查看列名
print("dfm_cov columns:", dfm_cov.columns.tolist())
print("dfh_cov columns:", dfh_cov.columns.tolist())
```

```python
# 将每个数据列（除了 chr, position）保存为 bedgraph.gz
# BedGraph 格式: chr, start, end, value

def save_bedgraph(df, prefix):
    """将每个值列保存为 bedgraph.gz"""
    value_cols = [c for c in df.columns if c not in ['chr', 'position']]
    for col in value_cols:
        bed = df[['chr', 'position']].copy()
        bed['end'] = bed['position'] + 1
        bed['value'] = df[col]
        bed = bed[['chr', 'position', 'end', 'value']]
        out_path = os.path.join(output_dir, f'{prefix}_{col}.bedgraph.gz')
        bed.to_csv(out_path, sep='\t', header=False, index=False, compression='gzip')
        print(f"Saved: {out_path}")

save_bedgraph(dfh_cov, '5hmC')
save_bedgraph(dfm_cov, '5mC')
print("Done!")
```
